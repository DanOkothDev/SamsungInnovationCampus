```python
def binary_search(S, x):
    low = 0
    high = len(S) - 1

    while low <= high:
        mid = (low + high) // 2

        if x == S[mid]:
            return mid
        elif x > S[mid]:
            low = mid + 1
        else:
            high = mid - 1

    return -1

S = [1, 2, 3, 4, 5, 6, 7, 8, 10]
x = 10

index = binary_search(S, x)

print(f"{x} is at position: {index}")

```
