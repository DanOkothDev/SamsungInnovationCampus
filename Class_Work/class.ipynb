```python
#Question 1

import math

class Circle:
    
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        
        return math.pi * self.radius * self.radius

    @property
    def circumference(self):
        
        return 2 * math.pi * self.radius

def main():
    rad = float(input("Enter the radius of the circle: "))
    circle = Circle(rad)

    print(f"The area of the circle is: {circle.area:.0f}")
    print(f"The circumference of the circle is: {circle.circumference:.0f}")

main()
```

```python
class Exam:
    def __init__(self, mark1, mark2, mark3):
        for mark in (mark1, mark2, mark3):
            if not 0 <= mark <= 100:
                raise ValueError("Each mark must be between 0 and 100!!")
                

            
        self.mark1 = mark1
        self.mark2 = mark2
        self.mark3 = mark3

    @property
    def average(self):
        return (self.mark1 + self.mark2 + self.mark3) / 3
        
    @property
    def grade(self):
        if 80 <= self.average <= 100:
            grade = "A"
        elif self.average >= 70:
            grade = "B"
        elif self.average >= 60:
            grade = "C"
        elif self.average >= 50:
            grade = "D"
        elif self.average >= 0:
            grade = "E"

        return grade

def main():
    score1, score2, score3 = map(float, input("Enter three marks separated by comas: ").split(","))

    sem1 = Exam(score1, score2, score3)

    print(f"The average score is: {sem1.average:.2f}, calculating to a grade {sem1.grade}")


main()
```

```python

class Grade:
    def __init__(self, score1, score2, score3):
        for score in (score1, score2, score3):
            if not 0 <= score <= 100:
                raise ValueError("Score must be between 0 and 100")
            
        self.score1 =score1
        self.score2 = score2
        self.score3 = score3

    @property
    def average(self):
        return (self.score1 + self.score2 + self.score3) / 3
        

    @property
    def get_grade(self):
        if 80 <= self.average <= 100:
            return "A"
        elif self.average >= 70:
            return "B"
        elif self.average >= 60:
            return "C"
        elif self.average >= 50:
            return "D"
        else:
            return "E"
            

def main():
    mark1, mark2, mark3 = map(float, input("Enter three marks, separated by coma: ").split(","))

    grade1 = Grade(mark1, mark2, mark3)

    print(f"Your grade is: {grade1.get_grade}")
    

main()
```

```python
class Small:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    @property
    def small(self):
        if self.a <= self.b and self.a <= self.c:
            return self.a
        elif self.b <= self.a and self.b <= self.c:
            return self.b
        else:
            return self.c
    
def main():
    num1, num2, num3 = map(int, input("Enter three numbers separated by space: ").split())

    small1 = Small(num1, num2, num3)

    print(f"Among, {num1}, {num2} and {num3} the smallest is {small1.small}")

main()
```

```python
class Even:
    def __init__(self, num):
        self.num = num

    @property
    def even(self):
        if self.num % 2 == 0:
            return 1
        else:
            return 0
    
def main():

    
    int_num = int(input("Enter a number: "))
    evens = Even(int_num)

    print(evens.even)

main()
```

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):

    @abstractmethod
    def area():
        pass

    @abstractmethod
    def get_dimension():
        pass
        
class Rectangle(Shape):
    def __init__(self):
        self.length = 0
        self.width = 0

    @property
    def area(self):
        return self.length * self.width

    @property
    def get_dimension(self):
        self.length, self.width = map(float, input("Enter the length and width of the rectangle: ").split())


class Triangle(Shape):
    def __init__(self):
        self.base = 0
        self.height = 0

    @property
    def area(self):
        return 0.5 * self.base * self.height

    @property
    def get_dimension(self):
        self.base, self.height = map(float, input("Enter the base and height of the triangle: ").split())

class Circle(Shape):
    def __init__(self):
        self.radius = 0

    @property
    def area(self):
        return math.pi * self.radius * self.radius

    @property
    def get_dimension(self):
        self.radius = float(input("Enter the radius of the circle: "))
        
def get_choice(choice):
    
    if choice == 1:
        triangle = Triangle()
        triangle.get_dimension
        print(f"The area of the 'Triangle' is: {triangle.area:.2f} cm^2")

    elif choice == 2:
        circle = Circle()
        circle.get_dimension
        print(f"The area of the 'Circle' is: {circle.area:.0f} cm^2")

    elif choice == 3:
        rectangle = Rectangle()
        rectangle.get_dimension
        print(f"The area of the 'Rectangle' is: {rectangle.area:.2f} cm^2")

    else:
        raise ValueError("Invalid input!!!")

def main():

    print("\nChoose a figure to calculate the area")
    print("1. Triangle")
    print("2. Circle")
    print("3. Rectangle\n")

    choice = int(input("Enter your choice(1-3): "))
    get_choice(choice)
    
    
main()
```

```python
class Temperature:
    def __init__(self, celcius):
        self.celcius = celcius

    
    def get_celcius(self):
        print(f"The current temperature is: {self.celcius}")

    @classmethod
    def from_fahrenheit(cls, f):
        c = (f - 32) * 5/9
        return cls(c)


    @staticmethod
    def is_freezing(celcius):
        if celcius <= 0:
            
            return True
        else:
            return False

temp1 = Temperature.from_fahrenheit(33)
print(Temperature.is_freezing(temp1.celcius))
temp1.get_celcius()
```

```python
class Student:

    def __init__(self, name, gpa, money):
        self.name = name
        self.gpa = gpa
        self.money = money
        self.songs = [
            "Song A",
            "Song B",
            "Song C"
        ]

    def __str__(self):
        return f"Student: {self.name}"

    def __repr__(self):
        return f"Student('{self.name}')"

    def __eq__(self, other):
        return self.name == other.name

    def __lt__(self, other):
        return self.gpa < other.gpa

    def __add__(self, other):
        return self.money + other.money

    def __len__(self):
        return len(self.songs)

student = Student("Boss", 2.4, 100)
student1 = Student("Boss", 3.4, 250)

print(student + student1)
```

```python
class Book:
    def __init__(self, title, pages, price):
        self.title = title
        self.pages =  pages
        self.price = price

    def __str__(self):
        return f"{self.title} ({self.pages} Pages)"

    def __repr__(self):
        return f"Book('{self.title}', {self.pages}, {self.price})"

    def __eq__(self, value):
        return self.title == value.title
   
    def __lt__(self, other):
        return self.pages < other.pages

    def __add__(self, other):
        return self.pages + other.pages

    def __len__(self):
        return self.pages

book1 = Book("Python Basics", 250, 19.99)
book2 = Book("System Design", 400, 29.99)

print(book1)
print(repr(book1))
print(book1 == book2)
print(book1 < book2)
print(book1 + book2)
print(len(book1))

    
```
