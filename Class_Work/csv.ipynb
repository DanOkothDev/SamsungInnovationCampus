```python
import csv

with open("students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "course", "year"])
    writer.writerow(["Dan", "BIT", 3])
    writer.writerow(["Amina", "BBIT", 2])
```

```python
with open("students.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

```python
with open("students.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row["name"], row["course"])
```

```python
with open("students.csv", "w", newline="") as f:
    fieldnames = ["name", "course", "year"]
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader() 
    writer.writerow({"name": "Dan", "course": "BIT", "year": 3})
    writer.writerow({"name": "Amina", "course": "BBIT", "year": 2})
```

```python
with open('units.csv', "w", newline="")as f:
    fieldnames = ["unit_code", "unit_name", "marks"]
    writer =csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerow({"unit_code": "SCT221", "unit_name":"IT", "marks": 90})
    writer.writerow({"unit_code": "SCT222", "unit_name":"BBIT", "marks": 75})
    writer.writerow({"unit_code": "SCT223", "unit_name":"BIT", "marks": 40})
    writer.writerow({"unit_code": "SCT224", "unit_name":"BBC", "marks": 85})
    writer.writerow({"unit_code": "SCT225", "unit_name":"HBC", "marks": 30})
    writer.writerow({"unit_code": "SCT226", "unit_name":"SBM", "marks": 48})

with open("units.csv", "r") as f:
    reader = list(csv.DictReader(f))
    #print(reader)
    
    high = list(filter(lambda s: int(s["marks"]) > 80, reader))
    for row in high:
        print(f"{row['unit_code']} - {row['unit_name']}: {row['marks']} marks")
```
