```python
import json

my_data = {"school": "JKUAT", "course": "BIT"}

# Convert to JSON string
json_text = json.dumps(my_data, indent=2)
print(json_text)
print(type(json_text))
```

```python
import json

my_data = {"school": "JKUAT", "course": "BIT"}

# dict → JSON string
json_text = json.dumps(my_data, indent=2)
print(json_text)
print(type(json_text))   

# JSON string → dict (back again)
restored_data = json.loads(json_text)
print(restored_data)
print(type(restored_data))   
```

```python
import json

def add_event(new_event):
    try:
        with open("events.json", "r") as f:
            events = json.load(f) 
            print(events)
              
    except FileNotFoundError:
        events = []   

    events.append(new_event)

    with open("events.json", "w") as f:
        json.dump(events, f, indent=2)

    


add_event({"title": "Baptism", "date": "04/08/2026", "location": "Nairobi"})
add_event({"title": "Youth Fellowship", "date": "10/08/2026", "location": "Annex Room"})
```
