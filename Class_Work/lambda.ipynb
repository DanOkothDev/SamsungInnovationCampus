```python
from dataclasses import dataclass

@dataclass
class Student:
    
    name: str
    age: int
    gpa : float = 4.0

    def __post_init__(self):
        if not 0 <= self.age <= 100:
            raise ValueError("Invalid age!!!")

student = Student("Dan", 100, 3.4)

print(student.age)
print(student.gpa)

```

```python


class Student:
    def __init__(self):
        self.name =  0
        self.age = 0
        self.gpa = 0

    @property
    def get_value(self):
        self.name = input("Enter your name: ")
        self.age = int(input("Enter your age: "))
        self.gpa = float(input("Enter your gpa: "))

    

def main():
    
    student1 = Student()
    student1.get_value

    print(student1.name)
    print(student1.age)
    print(student1.gpa)

if __name__ == "__main__":
    main()




```
