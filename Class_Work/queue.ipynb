```python
from collections import deque

class Queue:
    def __init__(self):
        self.value = deque()

    def enqueue(self, value):
        self.value.append(value)

    def dequeue(self):
        return self.value.pop()

    def is_empty(self):
        return len(self.value) == 0

    @staticmethod
    def add(a, b):
        return a + b

queue = Queue()


queue.enqueue(20)
queue.enqueue(40)
queue.dequeue()

print(queue.is_empty())
print(queue.value)
```

```python
class Deque:
    def __init__(self):
        self.queue = []

    def add_first(self, value):
        self.queue.insert(0, value)

    def remove_first(self):
        return self.queue.pop(0)

    def add_last(self, value):
        self.queue.append(value)

    def remove_last(self):
        return self.queue.pop()

deq = Deque()
deq.add_first(40)
deq.add_first(30)
deq.add_first(20)
deq.add_first(10)
deq.remove_first()
deq.add_last(45)
deq.add_last(60)
deq.remove_last()
print(deq.queue)
```

```python
queue = Queue()

items = [10 * i for i in range(1, 11)]

for item in items:
    queue.enqueue(item)
    if (item // 10) % 2 == 0:
        queue.dequeue()

print(queue.value)
```

```python
def chair_elimination_sequence(n, k):
    sequence = []
    queue = []

    for i in range(n + 1):
        queue.append(i)

    

    while len(queue) > 1:
        item = queue.pop(0)
        if i % k == 0:
            sequence.append(item)
        else:
            queue.append(item)

        i += 1

    sequence.append(queue.pop(0))
    return sequence

chair_elimination_sequence(7, 3)
```
