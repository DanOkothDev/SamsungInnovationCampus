```python
import math
import random

radii = [2, 4.5, 7, 1.2]

x = list(filter(lambda s : s > 3, radii))
areas = list(map(lambda r : math.pi * r * r, x))

rand = random.sample(areas, k=1)

print(f"{rand[0]:.2f}")
```

```python
from functools import reduce
# syntax reduce(function, iterable, initial_value)

numbers = [1, 2, 3, 4, 5]
total = reduce(lambda acc, n: acc + n, numbers, 0)
print(total)
```

```python
numbers = [1, 2, 3, 4, 5]
largest = reduce(lambda acc, n : acc if acc > n else n, numbers)
biggest = max(numbers)
print(biggest)
print(largest)
```

```python
from functools import reduce

questions = [
    {"text": "What is Big-O of binary search?", "difficulty": "medium", "marks": 5},
    {"text": "Define a hashmap.", "difficulty": "easy", "marks": 2},
    {"text": "Explain closures.", "difficulty": "hard", "marks": 10},
    {"text": "What is a list comprehension?", "difficulty": "easy", "marks": 3},
]


easy = list(filter(lambda q: q["difficulty"] == "easy", questions))

marks_only = list(map(lambda q: q["marks"], easy))

total_marks = reduce(lambda acc, m: acc + m, marks_only, 0)

print(f"Total marks available for easy questions: {total_marks:.2f}")
```
