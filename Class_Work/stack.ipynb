```python
def check(expr):
    opening = "[{("
    closing = "]})"

    stack = []
    for char in expr:
        if char in opening:
            stack.append(char)
        elif char in closing:
            if len(stack) == 0:
                return False

            if opening.index(stack.pop()) != closing.index(char):
                return False

    return len(stack) == 0

```

```python
str = input("Input a string of parentheses: ")
print("Valid" if check(str) else "Invalid")
```

```python
class Stack:
    def __init__(self):
        self.stack = []

    def is_empty(self):
        return True if len(self.stack) == 0 else False

    def push(self, item):
        self.stack.append(item)

    def pop(self):
        return None if self.is_empty() else self.stack.pop()

    def peek(self):
        return None if is_empty(self.stack) else self.stack[-1]
```

```python

import re


def extract_tags(html: str):
    
    # Matches things like <html>, </body>, <li>, </ul>
    tag_pattern = re.compile(r"<(/?)([a-zA-Z][a-zA-Z0-9]*)[^<>]*?(/?)>")

    tags = []
    for match in tag_pattern.finditer(html):
        is_closing = match.group(1) == "/"
        tag_name = match.group(2).lower()
        is_self_closing = match.group(3) == "/"

        if is_self_closing:
            continue  # e.g. <br/> needs no partner

        tags.append((tag_name, is_closing))

    return tags


def match_html_tags(html: str):
    
    stack = []
    tags = extract_tags(html)

    for tag_name, is_closing in tags:
        if not is_closing:
            # Opening tag -> push onto stack
            stack.append(tag_name)
        else:
            # Closing tag -> the stack must not be empty,
            # and the top of the stack must match this tag.
            if not stack:
                return False, f"Error: found </{tag_name}> with no matching opening tag."

            expected_tag = stack.pop()
            if expected_tag != tag_name:
                return False, (
                    f"Error: expected </{expected_tag}> but found </{tag_name}> instead."
                )

    if stack:
        unclosed = ", ".join(f"<{tag}>" for tag in stack)
        return False, f"Error: these tags were never closed: {unclosed}"

    return True, "All tags are correctly matched."


if __name__ == "__main__":
    html_doc = """
    <html>
    <body>
    <h1>Hello, World!</h1>
    <p> We are learning the art of coding
    with Python programming language.
    Here we are learning ... </p>
    <ul>
    <li> Data Structures, </li>
    <li> Algorithms, </li>
    <li> and Computational Thinking,
    eventually. </li>
    </ul>
    </body>
    </html>
    """

    is_valid, message = match_html_tags(html_doc)
    print(message)

    
```

```python
def extract_tags(html: str):
    
    tags = []
    position = 0  # where we currently are in the string

    while True:
        start = html.find("<", position)
        if start == -1:
            break  # no more tags in the document

        end = html.find(">", start)
        if end == -1:
            break  # malformed HTML: '<' with no matching '>'

        # Everything between < and > , e.g. "/ul" or "img src='x.png' /"
        inside = html[start + 1:end].strip()

        # Self-closing tag, e.g. <br/> or <img .../>
        if inside.endswith("/"):
            position = end + 1
            continue

        # Closing tag, e.g. </ul>
        if inside.startswith("/"):
            tag_name = inside[1:].strip().lower()
            tags.append((tag_name, True))
            position = end + 1
            continue

        # Opening tag, e.g. <ul> or <li class="item">
        # The tag name is just the first "word" before any space
        tag_name = inside.split(" ")[0].strip().lower()
        tags.append((tag_name, False))
        position = end + 1

    return tags


def match_html_tags(html: str):
    
    stack = []
    tags = extract_tags(html)

    for tag_name, is_closing in tags:
        if not is_closing:
            stack.append(tag_name)
        else:
            if not stack:
                return False, f"Error: found </{tag_name}> with no matching opening tag."

            expected_tag = stack.pop()
            if expected_tag != tag_name:
                return False, (
                    f"Error: expected </{expected_tag}> but found </{tag_name}> instead."
                )

    if stack:
        unclosed = ", ".join(f"<{tag}>" for tag in stack)
        return False, f"Error: these tags were never closed: {unclosed}"

    return True, "All tags are correctly matched."


if __name__ == "__main__":
    html_doc = """
    <html>
    <body>
    <h1>Hello, World!</h1>
    <p> We are learning the art of coding
    with Python programming language.
    Here we are learning ... </p>
    <ul>
    <li> Data Structures, </li>
    <li> Algorithms, </li>
    <li> and Computational Thinking,
    eventually. </li>
    </ul>
    </body>
    </html>
    """

    is_valid, message = match_html_tags(html_doc)
    print(message)
```

```python
html_doc = """
<html>
<body>
<h1>Hello, World!</h1>
<p>We are learning Python.</p>
<ul>
<li>Data Structures</li>
<li>Algorithms</li>
</ul>
</body>
</html>
"""

tags = []

i = 0

while i < len(html_doc):

    if html_doc[i] == "<":

        tag = ""

        while html_doc[i] != ">":
            tag += html_doc[i]
            i += 1

        tag += ">"

        tags.append(tag)

    i += 1

print(tags)
```

```python
stack = []

for tag in tags:

    if tag.startswith("</"):

        if len(stack) == 0:
            print("Invalid HTML")
            break

        opening = stack.pop()

        if opening[1:-1] != tag[2:-1]:
            print("Tags do not match!")
            break
        
    else:
        stack.append(tag)
else:
    if len(stack) == 0:
        print("HTML is valid.")

    else:
        print("Some tags were not closed.")
```
