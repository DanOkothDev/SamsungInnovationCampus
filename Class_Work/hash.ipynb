```python
table = {"I": 1, "V": 5, "X": 10, "L": 50, "C": 100, "D": 500, "M": 1000}

def roman_to_int(str):
    result = 0
    for i in range(len(str) - 1):

        if table[str[i]] < table[str[i + 1]]:
            result -= table[str[i]]

        else:
            result += table[str[i]]

    return result + table[str[-1]]


roman = input("Enter a roman number: ")
number = roman_to_int(roman)


print(number)

```

```python
books = [
    "The Little Prince",
    "The Old Man and the Sea",
    "The Little Mermaid",
    "Beauty and the Beast",
    "The Last Leaf"
]

for book in books:
    key = sum(map(ord, book))
    print(key, book)
```

```python
table = [None] * 10

def my_hash(key):
    return len(key) % 10

def insert(key, value):
    index = my_hash(key)
    table[index] = (key, value)

insert("Dan", 20)
insert("Alice", 21)
insert("Bob", 19)

print(table)
```

```python
table = [[] for _ in range(10)]

def my_hash(key):
    return len(key) % 10

def insert(key, value):
    index = my_hash(key)
    table[index].append((key, value))

insert("Dan", 20)
insert("Alice", 21)
insert("Bob", 19)

print(table)
```

```python

class BankAccount:
    def __init__(self, name, balance):
        self.__name = name
        self.__balance = balance

    @property
    def get_accounts_info(self):
        return f"{self.__name} has Ksh {self.__balance}"

    
    def deposit(self, amount):
        self.__balance  += amount
        return self.__balance

    
    def withdraw(self, amount):
        self.__balance -= amount
        return self.__balance

acc1 = BankAccount("Dan", 20000)
acc2 = BankAccount("Lucy", 30000)

print(acc1.get_accounts_info)
print(acc2.get_accounts_info)

acc1.deposit(40000)
print(acc1.get_accounts_info)
print(acc1._BankAccount__balance)

    
```

```python
from abc import ABC, abstractmethod

class Car(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def color(self):
        return "Blue"

class Benz(Car):
    def __init__(self, model, year):
        self.model = model
        self.year = year

    def describe(self):
        return f"{self.model}: {self.year}"

    def start(self):
        return f"{self.model} has started"

    def color(self):
        return super().color()

benz = Benz("GLE", 2026)
print(benz.model)
print(benz.year)
benz.color()
        
```
