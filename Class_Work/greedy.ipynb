```python
def greedy(amount, coins):
    coins_returned = []

    for i in range(len(coins)):
        while amount > coins[i]:
            amount -= coins[i]
            coins_returned.append(coins[i])

    return coins_returned


coins = [100, 50, 30, 20, 10]

change = greedy(1000, coins)


print(change)


        
```
