```python
import math

def area_circle(r):
    area = math.pi * r * r
    return area

def circumference_circle(r):
    circ = 2 * PI * r
    return circ

radius =  float(input("Enter the radius of the circle: "))

rad = area_circle(radius)
circumference = circumference_circle(radius)

print(f"Area: {rad}, Circumference: {circumference}")


    
```

```python
def area_rec(len, wid):
    area = len * wid
    return area

len, wid = map(float, input("Enter the length and width of the Rectangle: ").split())

print(f"The area of the rectangle is: {area_rec(len, wid)}")
```
