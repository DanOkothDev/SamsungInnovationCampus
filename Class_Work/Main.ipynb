```python
pin  = 1234
no_of_attempts = 0
attempts = 3
max_attempts = 3
i = 0

while(i < attempts):
    i += 1
    pinNo = int(input("Enter your pin: "))

    if no_of_attempts == 2 or max_attempts == 0:
            print("Card blocked due to many incorrect attempts")

    elif pinNo != pin:
        no_of_attempts += 1
        print(f"Incorrect pin, {max_attempts - 1} attempts remaining, try again!!") 
        max_attempts -= 1
        
            
    else:
        print("Welcome")
        break
   
```

```python
class Shape:
    def __init__(self, color, is_filled):
        self.color = color
        self.is_filled = is_filled

class Rectangle(Shape):
    def __init__(self, color, is_filled, length, width):
        super().__init__("Red", is_filled)
        self.length = length
        self.width = width
rect = Rectangle("Blue", True, 30, 40)
print(rect.color)
```
