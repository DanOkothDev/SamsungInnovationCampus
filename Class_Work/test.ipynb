```python
def bubblesort(S):
    n = len(S)
    no_swap = 0
    for i in range(n):
        print(S)
        for j in range(n - 1):
            if S[j] > S[j + 1]:
                no_swap += 1
                S[j], S[j + 1] = S[j + 1], S[j]

    print(f"Number of swaps is: {no_swap}")

S = [50, 30, 40, 10, 20]
bubblesort(S)

print(S)

```

```python
def insertionsort2(S):
    n = len(S)
    count = 0
    for i in range(1, n):
        print(S)
        x = S[i]
        j = i - 1
        while j >= 0 and S[j] > x:
            count += 1
            S[j + 1] = S[j]

            j -= 1
        S[j + 1] = x
    print(f"Number of comparisons: {count}")
S = [50, 40, 30, 10, 20]
insertionsort2(S)
print(S)
```

```python
count = 0

def merge2(S, low, mid, high):
    global count

    count += 1

    R = []
    i, j = low, mid + 1

    while i <= mid and j <= high:
        if S[i] < S[j]:
            R.append(S[i]); i += 1
        else:
            R.append(S[j]); j += 1

    if i > mid:
        for k in range(j, high + 1):
            R.append(S[k])
    else:
        for k in range(i, mid + 1):
            R.append(S[k])

    for k in range(len(R)):
        S[low + k] = R[k]


def mergesort2(S, low, high):
    
    if low < high:
        print(S)
        mid = (low + high) // 2
        mergesort2(S, low, mid)
        mergesort2(S, mid + 1, high)
        merge2(S, low, mid, high)
        

    

S = [6, 2, 11, 7, 5, 4, 8, 16, 10, 3]
mergesort2(S, 0, len(S) - 1)
print(S)
print(f"Merge2 called: {count} times")
```

```python
def bubblesortmine(S):
    n = len(S)

    for i in range(n):
        
        for j in range(n - 1):
            if S[j] > S[j + 1]:
                S[j], S[j + 1] = S[j + 1], S[j]

    return S

S = [5, 2, 4, 1, 3]
sorted = bubblesortmine(S)
print(sorted)
```

```python
def selectionsortmine(S):
    n = len(S)
    
    for i in range(n):
        smallest = i
        for j in range(i + 1, n):
            if S[j] < S[smallest]:
                smallest = j

        S[i], S[smallest] = S[smallest], S[i]

    return S

S = [5, 2, 4, 1, 3]
sorted = selectionsortmine(S)
print(sorted)
        
            
```

```python
def insertionsortmine(S):
    n = len(S)

    for i in range(1, n):
        key = S[i]
        j = i - 1

        while j >= 0 and S[j] > key:
            S[j + 1] = S[j]
            j -= 1

        S[j + 1] = key

    return S


S = [5, 2, 4, 1, 3]
sorted = insertionsortmine(S)

print(sorted)
```
