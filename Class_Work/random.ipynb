```python
import random

questions = [
    {"text": "What is Big-O of binary search?", "difficulty": "medium", "marks": 5},
    {"text": "Define a hashmap.", "difficulty": "easy", "marks": 2},
    {"text": "Explain closures.", "difficulty": "hard", "marks": 10},
    {"text": "What is a list comprehension?", "difficulty": "easy", "marks": 3},
    {"text": "Explain the loop-capture trap.", "difficulty": "hard", "marks": 10},
    {"text": "What is a dictionary?", "difficulty": "easy", "marks": 2},
    {"text": "Explain *args and **kwargs.", "difficulty": "medium", "marks": 6},
]

q = list(filter(lambda s : s["difficulty"] == "easy", questions))
x = random.sample(q, k=2)
total = sum(question["marks"] for question in x)
print(x)
print(f"Total Marks: {total:.2f}")
```
