```python
cases_3times = set()
for i in range(1, 7):
    for j in range(1, 7):
        for k in range(1, 7):
            cases_3times.add((i, j, k))

            total_cases = len(cases_3times)
print('Event that can happen by rolling the dice 3 times are', total_cases, 'cases.')
```

```python
for i in range(3, 19): #Find the probability of getting more than 3 to more than 18
    total_cases = len(cases_3times)
    n_cases = 0
    for c in cases_3times:
        if sum(c) >= i:
            n_cases += 1
    prob = n_cases * 100 / total_cases
    print(f"probability of getting more than {i:2d} is {prob:6.2f}%")
```

```python
A = {1, 3}    # element of set A
B = {2, 4}    # element of set B
res = set()

for i in A:
    for j in B:
        res = res | {(i, j)}    # product set with double for loop
AxB = res    # product set AxB of A and B(not A *B)
print('A =', A)
print('B =', B)
print('AxB =', AxB)    # print product set of A and B
```

### 2. Aggregation Using Zip Function

```python
empty_iterator = zip()
result = set(empty_iterator)
result
```

```python
a = [10, 20, 30]                   # list a
b = ('ten', 'twenty', 'thirty')    #tuple b
for val in zip(a, b):              # print tuple created by aggregating list a and tuple b
    print(val)
```

```python
lst = ['a', 'b', 'hello', 'this']
my_iterator = zip(lst)
result = set(my_iterator)
result
```

```python
str_list = ['hello', 'world', 'python', 'rocks']
int_tuple = (1, 2, 3)
int_list = [400, 500, 600, 700, 800]
my_iterator = zip(str_list, int_tuple, int_list)
list(my_iterator)
```

```python
zip_lst = list(zip(str_list, int_tuple, int_list))
a, b, c = zip(*zip_lst)    # unpack tuple
print(a, b, c)
```
