### 1. Implementing Merge Sort with Additional Memory

```python
def merge1(S, L, R):
    k = 0
    while len(L) > 0 and len(R) > 0:
        if L[0] <= R[0]:
            S[k] = L.pop(0)
        else:
            S[k] = R.pop(0)
        k += 1
    while len(L) != 0:
        S[k] = L.pop(0)
        k += 1
    while len(R) != 0:
        S[k] = R.pop (0)
        k += 1
```

```python
def mergesort1(S):
    n = len(S)
    if n > 1:
        print (S)
        mid = n // 2
        L, R = S[:mid], S[mid:]
        mergesort1(L)
        mergesort1(R)
        merge1(S, L, R)
```

```python
S = [27, 10, 12, 20, 25, 13, 15, 22]
mergesort1(S)
print(S)
```

### 2. Implementing Merge Sort More Efficiently

```python
def merge2(S, low, mid, high):
    R = []
    i, j = low, mid + 1
    while i <= mid and j <= high:
        if S[i] < S[j]:
            R.append(S[i]); i += 1
        else:
            R.append(S[j]); j += 1
    if i > mid:
        for k in range(j, high + 1):
            R.append(S[j])
    else:
        for k in range(i, mid + 1):
            R.append(S[i])
    for k in range(len (R)):
        S[low + k] = R[k]
```

```python
def mergesort2(S, low, high):
    if low < high:
        print(S)
        mid = (low + high) // 2
        mergesort2(S, low, mid)
        mergesort2(S, mid + 1, high)
        merge2(S, low, mid, high)
```

```python
S = [27, 10, 12, 20, 25, 13, 15, 22]
mergesort2(S, 0, len(S) - 1)
print(S)
```
