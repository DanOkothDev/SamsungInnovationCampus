## Let’s code

### 1. Swap Operation

```python
def swap1(S, x, y):
    t = S[x]
    S[x] = S[y]
    S[y] = t
```

```python
S = [10, 20]
print(S)
swap1(S, 0, 1)
print(S)
```

```python
def swap2(S, x, y):
    S[x], S[y] = S[y], S[x]
```

```python
S = [10, 20]
print(S)
swap2(S, 0, 1)
print(S)
```

### 2. Bubble Sort

```python
def bubblesort(S):
    n = len(S)
    for i in range(n):
        print(S)
        for j in range(n - 1):
            if S[j] > S[j + 1]:
                S[j], S[j + 1] = S[j + 1], S[j]
```

```python
S = [50, 30, 40, 10, 20]
bubblesort(S)
print(S)
```

### 3. Selection Sort

```python
def selectionsort1(S):
    R = []
    while len(S) > 0:
        print(R, S)
        smallest = S.index(min(S))
        R.append(S[smallest])
        S.pop(smallest)
    return R
```

```python
S = [50, 30, 40, 10, 20]
R = selectionsort1(S)
print(R)
```

```python
def selectionsort2(S):
    n = len(S)
    for i in range(n - 1):
        print(S)
        smallest = i
        for j in range(i + 1, n):
            if S[j] < S[smallest]:
                smallest = j
        S[i], S[smallest] = S[smallest], S[i]
```

```python
S = [50, 30, 40, 10, 20]
selectionsort2(S)
print(S)
```

```python
def insertionsort2(S):
    n = len(S)
    for i in range (1, n):
        print(S)
        x = S[i]
        j = i - 1
        while j >= 0 and S[j] > x:
            S[j + 1] = S[j]
            j -= 1
        S[j + 1] = x
```
