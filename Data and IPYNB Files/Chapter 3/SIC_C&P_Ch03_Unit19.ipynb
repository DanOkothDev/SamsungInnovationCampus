```python
def valid_date (date):
    days = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    sherlock_series = [1887, 1890, 1892, 1894, 1901, 1905, 1915, 1917, 1927]
    splitted = date.split('.')
    year = int(splitted[0])
    month = int(splitted[1])
    day = int(splitted [2])
    is_sherlock_book_yr = lambda x: x in sherlock_series

    if not is_sherlock_book_yr(year):
        print('This is not a publication year of the Sherlock Holmes series.')
        return False
    if month < 1 or month > 12:
        print('The month is incorrect.')
        return False
    if day < 1 or day > days [month-1]:
        print('The date is incorrect.')
        return False
    
    return True

date = input('Distinguish year, month and date by •(e.g. 1917.6.10) : ')
if valid_date(date) == True:
    print('This is the publication date of a Sherlock Holmes series!')
else:
    print('This is not the publication date of a Sherlock Holmes series or the input has an error.')
```

```python
# addition using add() function
def add(x, y) :
    return x + y

print('sum of 100 and 200 :', add (100, 200))
```

### 2. filter, map, reduce

```python
# return True for values over 19, and False for those that are not.
def adult_func(n):
    if n >= 19:
        return True
    else:
        return False

ages = [34, 39, 20, 18, 13, 54]
print('adults list :')
for a in filter (adult_func, ages): # filter ages by using filter () function
    print(a, end = ' ')
```

```python
ages = [34, 39, 20, 18, 13, 54]
print('adults list :')
for a in filter(lambda x: x >= 19, ages): # filter ages using filter () function
    print(a, end = ' ')
```

```python
a = [1, 2, 3, 4, 5, 6, 7]
square_a = []
for n in a:
    square_a.append (n**2) # add n square to list square a
print(square_a)
```

```python
def square(x):
    return x ** 2

a = [1, 2, 3, 4, 5, 6, 7]
# apply the return value of square function to each term of a
square_a = list(map(square, a))
print(square_a)
```

```python
def divide(x, y):
    try:
        result = x / y
    except ZeroDivisionError:
        print('error by zero division')
    else:
        print('result :', result)
    finally:
        print('execution complete')

print('divide (100, 2) function call :')
divide (100, 2)
print ('divide (100, 0) function call :')
divide (100, 0)
```

### 3. Comparison of List Comprehension and Lambda Expression

```python
a = [1, 2, 3, 4, 5, 6, 7]            # list of consecutive values
a = list(map(lambda x: x**2, a))     # apply lambda function for each element of the list
print(a)
```

```python
a = [1, 2, 3, 4, 5, 6, 7]    # list of consecutive values
a = [x**2 for x in a]        # apply x**2 for each element of the list
print(a)
```

```python
a = [x**2 for x in range(1, 8)]
print(a)
```

```python
ages = [34, 39, 20, 18, 13, 54]
adult_ages = list(filter(lambda x: x >= 19, ages))
print('adults list:', adult_ages)
```

```python
ages = [34, 39, 20, 18, 13, 54]
print('adults list:', [x for x in ages if x >= 19])
```
