```python
class BankAccount:
    """A class example in python that models a bank account"""
    def __init__(self, name, account_num, balance = 0):
        self.name = name
        self.account_num = account_num
        self.balance = balance
    
    def get_name(self):
        return self.name
    
    def get_account_num(self):
        return self.account_num
    
    def get_balance(self):
        return self.balance
    
    def deposit(self, amount):
        self.balance += amount
        print(f'{amount} won has been deposited. The balance is {self.balance:,} won.')
        return self.balance
    
    def withdraw(self, amount):
        if self. balance - amount > 0 :
            self.balance -= amount
        else:
            print(f'The account balance is {self.balance:,} won. '
                  f'This is smaller than the requested amount of withdrawal {amount} won.')
    
    def __str__(self):
        return f"The balance of {self.name}'s account {self.account_num} is {self.balance:,} won."

account1 = BankAccount ("David", "1234-0001")
print(account1)
account1.deposit(2000)
print(account1)
account1.withdraw(500)
print(account1)
account1.withdraw(5000)
```

### 6. Class Variables and Methods

```python
class Circle:
    def __init__(self, name, radius, PI):
        self.__name = name        # instance variable
        self.__radius = radius    # instance variable
        self.__PI = PI            # instance variable
    
    # Compute the area by multiplying radius**2 to current instance's PI
    def area(self):
        return self.__PI * self.__radius ** 2

c1 = Circle("C1", 4, 3.14)
print("Area of c1:", c1.area())
c2 = Circle ("C2", 6, 3.141)
print("Area of c2:", c2.area ())
c3 = Circle ("C3", 5, 3.1415)
print("Area of c3:", c3.area ())
```

```python
class Circle:
    PI = 3.1415    # class variable
    def __init__(self, name, radius):
        self.__name = name        # instance variable
        self.__radius = radius    # instance variable
    
    #Compute the area by using Circle’s class variable PI
    def area(self):
        return Circle.PI * self.__radius ** 2

c1 = Circle("C1", 4)
print("Area of c1:", c1.area())
c2 = Circle ("C2", 6)
print("Area of c2:", c2.area ())
c3 = Circle ("C3", 5)
print("Area of c3:", c3.area ())
```

### 7. Class Inheritance

```python
class A :
    PI = 3.14
    
class B(A):
    pass

a = A()    # Generates instance a of class A
b = B()    # Generates instance a of class B

print('a.PI =', a.PI)
print('b.PI =', b.PI)    # b can access the attributes of the parent class
```

```python
class Cat:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age
    
    # string expression format of Cat objects
    def __str__(self):
        return 'Cat (name='+self.__name+', age='+str(self.__age)+')'
    
    # limits random external access on self. age and prevents age from being negative
    def set_age(self, age):
        if age > 0:
            self.__age = age

    def get_age(self):
        return self.__age

nabi = Cat('nabi', 3)    # generates instance nabi
print(nabi)
nabi.set_age(4)          # approaches age via set aqe method
nabi.set_age(-5)         # prevents age from being negative via set age() method
print(nabi)
```

### 5. Class and Inheritance

```python
class Person:
    def __init__(self, name):
        self.name = name
    def get_name(self):
        return self.name

class Employee(Person):    # child class of the Person class
    def __init__(self, name, staff_id):
        Person.__init__(self, name)
        self.staff_id = staff_id
    def info(self):
        return 'staff : ' + self.get_name() + ', staffID : ' + str(self.staff_id)

class Manager (Person):    # child class of the Person class
    def __init__(self, name, position):
        Person.__init__(self, name)
        self.position = position
    def info(self):
        return 'Manager : ' + self.get_name() + ', position : ' + str(self.position)

worker1 = Employee('David Doe', 1111)
worker2 = Employee('Paul Carter', 2222)
cfo = Manager ('Anna Smith', 'CFO')

print(worker1.info())
print(worker2.info())
print(cfo.info())
```
