```python
def urban (city):
    global max_pop, min_pop, pop_sum
    n = 0
    for name, pop in city.items(): # obtains maximum and minimum value by iterating in the loop statement
        if pop > max_pop:
            max_pop = pop
        if pop < min_pop:
            min_pop = pop
        pop_sum += pop
        n += 1
    print ('maximum population: ', max_pop)
    print('minimum population: ', min_pop)
    print('difference between maximum population and minimum population:', max_pop - min_pop)
    print ('average population:', pop_sum / n)

max_pop = 0
min_pop = 1000000
pop_sum = 0
# stores population statistics (unit : thousands) data in a dictionary
city_pop = { 'A': 9765, 'B':3441, 'C': 2954, 'D': 1531 }
urban(city_pop)
```

### 1. Python’s scoping rule

```python
x = 10
y = 11            # corresponds to G
def foo():
    x = 20        # foo function corresponds to L
    def bar():
        a = 30    # bar function corresponds to E
        print(a, x, y) # each variable corresponds to L, E and G
    bar()
    x = 40
    bar()

foo()
```

### 2. Outer Function and Nested Function

```python
def decorate (style = 'italic'):
    def italic(s):
        return '<i>' + s + '</i>'
    def bold(s):
        return '<b>' + s + '</b>'
    if style == 'italic':
        return italic
    else:
        return bold
dec = decorate()
print(dec('hello'))
dec2 = decorate('bold')
print(dec2('hello'))
```

```python
def another_func():
    print("hello")

def outer_func():
    return another_func()

outer_func()
```

### 3. Process of Finding Nonlocal Variable

```python
n1 = 1
def func1():
    def func2():
        global n1
        n1 += 1
        print(n1) # print 2
    func2()

func1()
```

```python
def func1():
    n2 = 1
    def func2():
        global n2
        n2 += 1
        print(n2) # error
    func2()

func1()
```

```python
def func1():
    n3 = 1
    def func2():
        nonlocal n3
        n3 += 1
        print(n3) # It's not error
    func2()

func1()
```

```python
x = 20 # global variable
def f():
    x = 40
    def g():
        nonlocal x
        x = 80
    g()
    print(x)

f()
print(x)
```

```python
X = 70 # global variable
def f():
    nonlocal x
    x = 140

f()
print(X)
```

```python
def f():
    a = 777
    def g():
        a = 100
        def h():
            nonlocal a
            a = 333
        h()
        print(f"[Level 2] a = {a}")
    g()
    print(f"[Level 1] a = {a}")

f()
```

### 4. Concept of Closure

```python
def closure_calc():
    a = 2
    def mult(x):
        return a * x
    return mult

c = closure_calc()
print (c(1), c(2), c(3))
```

```python
def makecounter() :
    count = 0
    def counter():
        nonlocal count
        count += 1
        return count
    return counter
c1 = makecounter()
c2 = makecounter()
print('c1', c1())
print('cl', c1())
print('c2', c2())
```

```python
def calc():
    a = 3
    b = 5
    def mul_add(x):
        return a * x + b
    return mul_add

c = calc()
print(c(1), c(2), c(3), c(4), c(5))
```

### 5. Making Closure with Lambda

```python
def closure_calc():
    a = 3
    b = 5
    return lambda x : a * x + b # making closure with lambda

c = clouser_calc()
print (c(1), c(2), c(3), c(4), c(5)) # prints 8, 11, 14, 17, 20
```

### 6. Change Local Variable of Closure

```python
def calc():
    a = 2
    b = 3
    total = 0
    def mult_add(x):
        nonlocal total
        total = total + a * x + b
        return total
    return mult_add

c = calc()
print(c(1), c(2), c(3))
```
