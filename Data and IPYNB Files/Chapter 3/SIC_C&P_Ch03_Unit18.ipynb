```python
def hanoi(n, from_, to_, via_):
    if n == 1:
        print(f"Number {n}, {from_} -> {to_}")
    else:
        hanoi(n - 1, from_, via_, to_)
        print(f"Number {n}, {from_} -> {to_}")
        hanoi(n - 1, via_, to_, from_)
hanoi(3, 'A', 'C', 'B')
```

## Key concept

### 1. Finding the Factorial with a Loop Statement

```python
n = int(input ('Enter a number : '))
fact = 1
for i in range(1, n+1):
    fact = fact * i

print(f"{n}! = {fact}")
```

### 2. Finding the Factorial with a Recursive Function

```python
def factorial(n):    # Implement recursive function of n!
    if n <= 1:       # Termination condition is required
        return 1
    else:
        return n * factorial (n-1)    # n * (n-1)! implementation by definition

n = 5
print(f"{n}! = {factorial(n)}")
```

### 3. Disadvantages of Recursive Functions and Memoization

```python
def fibonacci (n):        # A recursive implementation of the Fibonacci function
    if n <= 1:            # Termination condition of the Fibonacci function
        return n
    else:
        return(fibonacci(n-1) + fibonacci(n-2))  # F_n = F_ (n-1) + F_ (n-2)

nterms = int(input("How many Fibonacci numbers do you want?"))

# Cannot find Fibonacci number if it is negative
if nterms <= 0:
    print("Error : Enter a positive number.")
else:
    print("Fibonacci sequence: ", end=' ')
    for i in range(nterms):
        print(fibonacci(i), end=' ')
```

```python
def fibonacci (n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(10))
```

```python
from timeit import *

def fibonacci_1(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci_1(n-1) + fibonacci_1(n-2)
    
t3 = Timer("fibonacci_1(30)", "from __main__ import fibonacci_1")
print("fibonacci_1(30) * 20 times : ", t3.timeit(number = 20), "seconds")
```

```python
from timeit import *    # Import all classes and methods of timeit

dic = {0:0, 1:1}
def fibonacci_1(n):
    if n in dic:
        return dic[n]
    dic[n] = fibonacci_1(n-1) + fibonacci_1(n-2)
    return dic[n]
    
t3 = Timer("fibonacci_1(30)", "from __main__ import fibonacci_1")
print("fibonacci_1(30) * 20 times : ", t3.timeit(number = 20), "seconds")
```
