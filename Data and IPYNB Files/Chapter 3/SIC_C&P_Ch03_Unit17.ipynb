### 4. Arbitrary Parameter

```python
def greet (*names):
    for name in names:
        print ('Hello' , name, '!')

greet('A', 'B', 'C')        # 3 arguments
greet ('James', 'Thomas')   # 2 arguments
```

```python
def foo(*args):
    print('The number of arguments:', len(args))
    print('Arguments:', args)

foo(10, 20, 30)
```

```python
def sum_nums(*numbers):
    result = 0
    for n in numbers:
        result += n
    return result

print(sum_nums(10, 20, 30))            # Print the sum of arguments of 10, 20, 30
print(sum_nums(10, 20, 30, 40, 50))    # Print the sum of arguments of 10, 20, 30, 40, 50
```

### 5. Default Parameter

```python
def print_star(n):        # One argument required
    for _ in range (n):
        print('*************************')

print_star () # An error occurs because there is no argument
```

```python
def print_star(n = 1):    # The parameter n has a default value of 1
    for _ in range (n):
        print('*************************')
        
print_star() # Execute without error even if there is no argument
```

```python
def print_star(n=1):   # The parameter n has a default value of 1
    for _ in range(n):
        print('***********************')

print_star(2)  # Since the argument value is 2, the default parameter n=1 is not performed
```

### 6. Keyword Parameter

```python
def get_root(a, b, c) :
    r1 = (-b + (b ** 2 - 4 * a * c) ** 0.5) / (2 * a)
    r2 = (-b - (b ** 2 - 4 * a * c) ** 0.5) / (2 * a)
    return r1, r2

# When calling a function, use a constant value as an argument
# Return the result value using result 1 and result 2
result1, result2 = get_root(1, 2, -8)
print('The result value is', result1, 'or', result2)
```

### 3. Multiple Return Statement

```python
def circle_area_circum(radius):
    area = 3.14 * radius ** 2
    circum = 2 * 3.14 * radius
    return area, circum # Returns a tuple - the return value (area, circum)

radius = 10
area, circum = circle_area_circum(radius) # Return the area and perimeter of a circle
print(f"The area of a circle of radius {radius} is {area:.1f}, "
      f"and the circumference of the circle is {circum:.1f})")
```

### 6. Product Set

```python
def product_set(set1, set2) :
    res = set()
    for i in set1:
        for j in set2:
            res = res | {(i, j)} # Product set using double for loop
    return res

def exp(input_set, exponent) : # A function that performs powers on input_set
    res = input_set # res initialization
    for _ in range(exponent-1) : # Repeated for (exponent-1) to become a power
        res = product_set(res, input_set)
    return res

A = {1, 3}
A3 = exp(A, 3) # Perform 3 powers on set A
print(A3)
```

```python
def product_set(set1, set2):
    res = set()
    for i in set1:
        for j in set2:
            res = res | {(i, j)}    # Product set using double for loop
    return res

cases = { 1, 2, 3, 4, 5, 6 }        # Integers from 1 to 6 on the die
cases_2times = product_set(cases, cases) # Get product_set for case
print(cases_2times)
```
