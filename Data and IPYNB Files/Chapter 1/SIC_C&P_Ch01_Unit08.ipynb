### 3. Mission

```python
breads = ["Rye bread", "Wheat", "White"]
vegis = ["Lettuce", "Tomato", "Cucumber"]
sauces = ["Mustard", "Tomato sauce", "Olive oil"]

# Religious dietary preferences mapped to protein options
religious_proteins = {
    "Muslim": ["Chicken patty", "Tofu patty", "Mushroom patty"],   # pork excluded
    "Buddhist": ["Tofu patty", "Mushroom patty"],                  # vegetarian-friendly
    "Christian/Others": ["Chicken patty", "Mushroom patty", "Tofu patty"]
}

print("Sandwich combinations considering religious dietary preferences\n")
for religion, proteins in religious_proteins.items():
    print(f"[{religion}] options:")
    for b in breads:
        for p in proteins:
            for v in vegis:
                for s in sauces:
                    print(f"  {b} + {p} + {v} + {s}")
    print()
```
