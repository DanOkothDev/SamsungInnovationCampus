### 1. Dynamic Programming

```python
def fib1(n):
    if n == 0 or n == 1:
        return n
    else:
        return fib1(n - 1) + fib1(n - 2)
```

```python
F = {0: 0, 1: 1}

def fib2(n):
    if n in F:
        return F[n]
    else:
        F[n] = fib2(n - 1) + fib2(n - 2)
        return F[n]
```

```python
def fib3(F, n):
    if n <= 1:
        return F[n]
    else:
        if F[n] < 0:
            F[n] = fib3(F, n - 1) + fib3 (F, n - 2)
        return F[n]
```

```python
N = int(input("Input a number: "))
F = [0, 1] + [-1] * (N - 1)
print(fib3(F, N))
```

```python
def bin1(n, k):
    if k == 0 or n == k:
        return 1
    else:
        return bin1(n - 1, k - 1) + bin1(n - 1, k)
```

```python
n = int(input("Input the value of n: "))
k = int(input("Input the value of k: "))
print(f"binomial({n}, {k}) is {bin1(n, k)}")
```

```python
def bin2(n, k):
    B = [[0] * (i + 1) for i in range(n + 1)]
    for i in range(n + 1):
        for j in range(i + 1):
            if j == 0 or j == i:
                B[i][j] = 1
            else:
                B[i][j] = B[i - 1][j - 1] + B[i - 1][j]
    return B[n][k]
```

```python
n = int(input("Input the value of n: "))
k = int(input("Input the value of k: "))
print(f"binomial({n}, {k}) is {bin2(n, k)}")
```

## Pop quiz

### Q1

```python
def fib4(n):
    if n <= 1:
        return n
    else:
        a, b = 0, 1
        for i in range(2, n + 1):
            a, b = b, a + b
        return b
```

```python
N = int(input("Input a number: "))
print(fib4(N))
```

### Q2

```python
def bin3(n, k):
    B = [0] * (n + 1)
    for i in range(n + 1):
        for j in range(n, -1, -1):
            if j == 0 or j == i:
                B[j] = 1
            else:
                B[j] = B[j] + B[j - 1]
    return B[k]
```

```python
for i in range (10):
    for j in range(i + 1):
        print(bin3(i, j), end=" ")
    print()
```
