### 8. Moving Statistics Function (Rolling Window Calculations)

```python
import pandas as pd
import datetime
from datetime import date, datetime, time
import matplotlib.pyplot as plt

import yfinance as yf
start = datetime(2025, 1, 1)
end = datetime(2025, 11,30)

sec = yf.download(tickers=['005930.KS'],
                  start=start,
                  end=end,
                  auto_adjust=False)
sec.head()
```

```python
sec.info()
```

```python
ma5 = sec['Adj Close'].rolling(window=5).mean()
ma5.head(10)
```

```python
close = sec['Adj Close']

ma5 = sec['Adj Close'].rolling(window=5).mean()
ma10 = sec['Adj Close'].rolling(window=10).mean()
ma60 = sec['Adj Close'].rolling(window=60).mean()
ma120 = sec['Adj Close'].rolling(window=120).mean()

plt.figure(figsize = (15,10))

plt.plot(close, label='close', linewidth=4)
plt.plot(ma5, label='5 window')
plt.plot(ma10, label='10 window')
plt.plot(ma60, label='60 window')
plt.plot(ma120, label='120 window')

plt.legend ()
plt.title('Rolling window calculations')
plt.xlabel('Date')
plt.ylabel('close price')
```

## Let’s code

```python
import pandas as pd
import numpy as np
import datetime
import matplotlib.pyplot as plt

from datetime import date, datetime, time, timezone

df = pd.read_csv("./data/Monthly Value/Monthly_Value.csv")
print(df.head())
print()
print(df.tail())
```

```python
df.info()
```

```python
# Search NaN data. You can check the number of missing data for each column.
df.isnull().sum()
```

```python
change_value=0
df = df.fillna(change_value)  
df.isnull().sum()
```

```python
df['Period'] = pd.to_datetime(df['Period'])
df.set_index('Period', inplace=True)

print(df.head())
print()
print(df.tail())
```

```python
df.info()
```

```python
plt.style.use('ggplot')
cols = ['Revenue', 'Average_cost']
df[cols].plot(kind='area',
              alpha=0.2,
              stacked=False,
              figsize=(20, 10))
plt.xlabel('Period')
plt.ylabel('Value')
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = df.index
y1 = df['Revenue']
y2 = df['Sales_quantity']
y3 = df['Average_cost']

fig, axs = plt.subplots(1, 3, figsize=(18, 5))

axs[0].plot(x, y1, color='blue')
axs[0].set_title('Revenue Over Time')
axs[0].set_xlabel('Period')
axs[0].set_ylabel('Revenue')

axs[1].plot(x, y2, color='green')
axs[1].set_title('Sales Quantity Over Time')
axs[1].set_xlabel('Period')
axs[1].set_ylabel('Sales Quantity')

axs[2].plot(x, y3, color='red')
axs[2].set_title('Average Cost Over Time')
axs[2].set_xlabel('Period')
axs[2].set_ylabel('Average Cost')

fig.suptitle('Monthly Performance Trends', fontsize=16)

plt.tight_layout()
plt.show()
```
