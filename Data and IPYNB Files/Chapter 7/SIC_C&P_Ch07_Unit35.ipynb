### 9. Basic Arithmetic Operation for Series Object (+, -, /, *, ...)

```python
import pandas as pd

s = [10,20,30,40,50]
t = [23,345,123]

sr1 = pd.Series(s)
sr2 = pd.Series(t)

sr3 = sr1 + sr2

print(sr3)
```

### 10. Pandas Provides Many Accessible Descriptive Statistics

```python
import pandas as pd

s = [10, 20,30,40,50]
t = [23,345,123]
sr1 = pd.Series(s)
sr2 = pd.Series(t)

sr3 = sr1 + sr2

print(sr3)

print(sr3.count())
print(sr3.sum())
```

## Let’s code

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
s_2024 = pd.Series(population_2024)
s_2025 = pd.Series(population_2025)
s_area = pd.Series(city_area)
```

```python
growth = s_2024 - s_2025
print(growth)
```

```python
x=growth.index
y=growth.values
```

```python
plt.figure(figsize = (13, 6))

plt.xticks(rotation='vertical', size = 10)

plt.barh(x, y)

plt.show()
```

```python
population_density = s_2024 / s_area
population_density
```

```python
x=population_density.index
y=population_density.values
```

```python
plt.figure(figsize = (13,6))

plt.xticks(rotation=45, size = 10)

plt.barh(x, y, color='red')

plt.title('population density', size = 15)
plt.ylabel('city', size = 15)
plt.xlabel('population', size = 15)
plt.tight_layout() 
plt.show()
```

```python
population_density.mean()
```
