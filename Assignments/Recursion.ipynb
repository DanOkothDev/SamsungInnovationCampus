```python
def calculate_sum(n):

    if n == 1:
        return 1

    return n + calculate_sum(n - 1)

print(calculate_sum(100))
```

```python
def get_sum(start, stop):
    if stop == start:
        return 1

    return stop + get_sum(start, stop - 1)

def main():
    n, m = map(int,input("Enter the start and stop values: ").split())
    print(get_sum(n,m))

if __name__=="__main__":
     main()
```
