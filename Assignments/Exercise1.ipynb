```python
#Question 07-01

adult = int(input("Are you an adult? (1 if youre an adult, 0 if youre minor): "))
if adult == 1 or adult == 0:
    married = int(input("Are you married? (1 if youre married, 0 if youre single): "))

    if adult and married:
        print("Youre an adult who is married")
    elif adult and not married:
        print("Youre an adult who is single")
    elif not adult and not married:
        print("You are a minor who is not married")
    elif not adult and married:
        print("You are a minor who is married")

else:
    print("Invalid input!!!!")
```

```python
#Question 08-01

for i in range(2, 13):
    is_prime = True

    for j in range(2, i):
        if i % j == 0:
            is_prime = False
            break
    if is_prime:
        print(f"{i} : Prime Number")
    else:
        print(f"{i} : Composite Number")
```

```python
#Question 09-01

for i in range(100, 1000):
    total = 0
    digit = str(i)
    for j in digit:
        total += (int(j) ** 3)
    
    if total == i:
        print(total)
```
