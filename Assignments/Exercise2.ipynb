```python
# Q. 12-01

lst = [5, 6, 3, 9, 2, 12, 3, 8, 7]

for i in range(len(lst) - 1):
    if lst[i] > lst[i + 1]:
        lst[i + 1], lst[i] = lst[i], lst[i + 1]

print(lst)
```

```python
# Q. 13-01

my_array = [[1, 2], [3, 4], [5, 6]]
new_array = []
for i in my_array:
    for j in i:
        new_array.append(j)

print(new_array)
```

```python
# Q. 14-01

maria = {"korean": 94, "english": 91, "mathematics": 89, "science": 83}
total = 0

for i in maria.values():
    total  += i

avg = total / len(maria)
print(avg)
 


```
